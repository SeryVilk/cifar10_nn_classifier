```python
import os

os.environ['TF_ENABLE_ONEDNN_OPTS'] = '0'

import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Conv2D, MaxPooling2D, Flatten, Dropout, BatchNormalization
from tensorflow.keras.utils import to_categorical

gpus = tf.config.list_physical_devices("GPU")
for gpu in gpus:
    tf.config.experimental.set_memory_growth(gpu, True)


def predict_single_image(model, image, class_names):
    """Предсказание для одного изображения"""
    image_batch = np.expand_dims(image, axis=0)
    prediction = model.predict(image_batch, verbose=0)
    predicted_class = np.argmax(prediction[0])
    confidence = np.max(prediction[0])

    return (class_names[predicted_class], confidence, prediction[0])


def get_classnames():
    """Возвращает названия классов CIFAR-10"""
    return ['airplane', 'automobile', 'bird', 'cat', 'deer',
            'dog', 'frog', 'horse', 'ship', 'truck']


def get_dataset():
    """Загрузка датасета CIFAR-10 из Keras"""
    from tensorflow.keras.datasets import cifar10

    # Загрузка данных
    (x_train, y_train), (x_test, y_test) = cifar10.load_data()

    # Преобразуем метки из 2D в 1D
    y_train = y_train.flatten()
    y_test = y_test.flatten()

    return ((x_train, y_train), (x_test, y_test))


def main():
    # Загрузка данных
    print("=" * 80)
    print("Загрузка датасета CIFAR-10...")
    (x_train, y_train), (x_test, y_test) = get_dataset()

    # Нормализация данных
    x_train = x_train.astype('float32') / 255.0
    x_test = x_test.astype('float32') / 255.0

    # One-hot кодирование меток
    y_train_cat = to_categorical(y_train, 10)
    y_test_cat = to_categorical(y_test, 10)

    # Получение названий классов
    class_names = get_classnames()
    print(f"Классы: {class_names}")
    print(f"Размер тренировочной выборки: {x_train.shape}")
    print(f"Размер тестовой выборки: {x_test.shape}")

    # Создание модели
    model = Sequential([
        Conv2D(32, 3, activation='relu', padding='same', input_shape=(32, 32, 3)),
        BatchNormalization(),
        Conv2D(32, 3, activation='relu', padding='same'),
        BatchNormalization(),
        MaxPooling2D((2, 2)),
        Dropout(0.25),

        Conv2D(64, 3, activation='relu', padding='same'),
        BatchNormalization(),
        Conv2D(64, 3, activation='relu', padding='same'),
        BatchNormalization(),
        MaxPooling2D((2, 2)),
        Dropout(0.25),

        Conv2D(128, 3, activation='relu', padding='same'),
        BatchNormalization(),
        Conv2D(128, 3, activation='relu', padding='same'),
        BatchNormalization(),
        MaxPooling2D((2, 2)),
        Dropout(0.25),

        Conv2D(256, 3, activation='relu', padding='same'),
        BatchNormalization(),
        Conv2D(256, 3, activation='relu', padding='same'),
        BatchNormalization(),
        MaxPooling2D((2, 2)),
        Dropout(0.25),

        Flatten(),
        Dense(512, activation='relu'),
        Dropout(0.5),
        Dense(256, activation='relu'),
        Dropout(0.5),
        Dense(10, activation='softmax')
    ])

    LAMBDA = 0.001
    model.compile(optimizer=tf.keras.optimizers.Adam(learning_rate=LAMBDA),
                  loss='categorical_crossentropy',
                  metrics=['accuracy'])

    model.summary()

    print("\n" + '=' * 100 + '\nОбучение нейросети')

    callbacks = [
        tf.keras.callbacks.EarlyStopping(
            monitor='val_loss',
            patience=5,
            restore_best_weights=True
        ),
        tf.keras.callbacks.ReduceLROnPlateau(
            monitor='val_loss',
            factor=0.5,
            patience=3,
            min_lr=1e-6
        )
    ]

    history = None
    if os.path.isfile('my_model.keras'):
        model = tf.keras.models.load_model('my_model.keras')
    else:
        history = model.fit(x_train, y_train_cat,
                            validation_data=(x_test, y_test_cat),
                            epochs=15,
                            callbacks=callbacks,
                            batch_size=64,
                            verbose=1)

        model.save('my_model.keras')
        print("Модель сохранена как 'my_model.keras'")

    print("\n" + '=' * 100 + '\nТочность нейросети и ошибка')


    if history is not None:
        # Графики обучения
        plt.figure(figsize=(12, 4))

        # График потерь
        plt.subplot(1, 2, 1)
        plt.plot(history.history['loss'], label='Training Loss')
        plt.plot(history.history['val_loss'], label='Validation Loss')
        plt.title('Loss During Training')
        plt.xlabel('Epoch')
        plt.ylabel('Loss')
        plt.legend()
        plt.grid(True)

        # График точности
        plt.subplot(1, 2, 2)
        plt.plot(history.history['accuracy'], label='Training Accuracy')
        plt.plot(history.history['val_accuracy'], label='Validation Accuracy')
        plt.title('Accuracy During Training')
        plt.xlabel('Epoch')
        plt.ylabel('Accuracy')
        plt.legend()
        plt.grid(True)

        plt.tight_layout()
        plt.savefig('training_history.png')
        plt.show()

    # Оценка модели
    loss, accuracy = model.evaluate(x_test, y_test_cat, batch_size=32, verbose=1)
    print(f"Test Loss: {loss:.4f}")
    print(f"Test Accuracy: {accuracy:.4f}")

    print("\n" + '=' * 100 + '\nПримеры ответов обученной нейросети')

    # Подготовка правильных и неправильных предсказаний
    correct = []
    incorrect = []

    # Используем тестовые данные для демонстрации
    for i in range(min(1000, len(x_test))):  # Проверяем только первые 1000 изображений
        image = x_test[i]
        true_label = y_test[i]

        # Предсказание
        pred_class, confidence, _ = predict_single_image(model, image, class_names)
        true_class = class_names[true_label]

        if pred_class == true_class and len(correct) < 4:
            correct.append((image, true_label))
        elif pred_class != true_class and len(incorrect) < 4:
            incorrect.append((image, true_label))

        if len(correct) == 4 and len(incorrect) == 4:
            break

    # Отображение примеров
    plt.figure(figsize=(12, 8))

    for i in range(8):
        plt.subplot(2, 4, i + 1)

        image_arr = correct
        j = i
        if i > 3:
            j = i - 4
            image_arr = incorrect

        plt.imshow(image_arr[j][0], interpolation='none')

        # Получаем предсказание
        pred_class, confidence, _ = predict_single_image(model, image_arr[j][0], class_names)
        true_class = class_names[image_arr[j][1]]

        # Определяем цвет заголовка
        color = 'green' if pred_class == true_class else 'red'

        plt.title(f"True: {true_class}\nPred: {pred_class}\nConf: {confidence:.2f}",
                  color=color, fontsize=10)
        plt.axis('off')

    plt.tight_layout()
    plt.savefig('predictions_examples.png')
    plt.show()

    # Дополнительно: предсказания на случайных изображениях
    print("\n" + '=' * 100)
    print("Предсказания на случайных изображениях из тестовой выборки:")

    # Выбираем 5 случайных индексов
    random_indices = np.random.choice(len(x_test), 5, replace=False)

    for idx in random_indices:
        image = x_test[idx]
        true_label = y_test[idx]

        pred_class, confidence, full_prediction = predict_single_image(model, image, class_names)
        true_class = class_names[true_label]

        print(f"\nИзображение {idx}:")
        print(f"  Истинный класс: {true_class}")
        print(f"  Предсказанный класс: {pred_class}")
        print(f"  Уверенность: {confidence:.4f}")
        print(f"  Правильно: {'Да' if pred_class == true_class else 'Нет'}")


if __name__ == "__main__":
    main()
```
